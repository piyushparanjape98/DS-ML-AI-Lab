```python
class MPNeuron:
  def __init__(self, weights, threshold):
    self.threshold = threshold
    self.weights = weights

  def predict(self, x1, x2):
    total = 0
    for w, x in zip(self.weights, [x1, x2]):
      total += w * x
    if total >= self.threshold:
      return 1
    return 0


And_Gate = MPNeuron([1,1], 2)                     # here the [1,1] are weights and 2 is threshold for and gate
inputs = [(0,0), (0,1), (1,0), (1,1)]
print("AND GATE")
for i in range(len(inputs)):
  print(f"Input: {inputs[i]} Output: {And_Gate.predict(inputs[i][0], inputs[i][1])}")   # here predicts - inputs[i][0] gets one element and [i][1] gets other

Or_Gate = MPNeuron([1,1], 1)
print('\n OR GATE')
for i in range(len(inputs)):
  print(f"Input: {inputs[i]} Output: {Or_Gate.predict(inputs[i][0], inputs[i][1])}")

NAND_Gate = MPNeuron([-1,-1], -1)
print('\n NAND GATE')
for i in range(len(inputs)):
  print(f"Input: {inputs[i]} Output: {NAND_Gate.predict(inputs[i][0] , inputs[i][1])}")

NOR_Gate = MPNeuron([-1,-1], 0)
print('\n NOR GATE')
for i in range(len(inputs)):
  print(f"Input: {inputs[i]} Output: {NOR_Gate.predict(inputs[i][0] , inputs[i][1])}")

NOT_Gate = MPNeuron([-1], 0)
print('\n NOT GATE')
for i in range(len(inputs)):
  print(f"Input: {inputs[i]} Output: {NOT_Gate.predict(inputs[i][0] , inputs[i][1])}")
```

The XOR gate cannot be implemented using a single McCulloch-Pitts (MP) neuron because it is not linearly separable. A single MP neuron can only create one linear decision boundary (a straight line) to separate the input classes. In the XOR truth table, the output is 1 for inputs (0,1) and (1,0), and 0 for (0,0) and (1,1). Since the inputs producing the same output lie on opposite corners, no single straight line can separate the 1s from the 0s. Therefore, no combination of weights and threshold can correctly implement the XOR function using a single MP neuron. To implement XOR, multiple neurons arranged in a multi-layer network (MLP) are required.
