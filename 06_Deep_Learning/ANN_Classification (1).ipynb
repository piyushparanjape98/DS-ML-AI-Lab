## DEEP LEARNING - ANN Classification

```python
import numpy as npimport pandas as pdfrom sklearn.model_selection import train_test_splitfrom sklearn.preprocessing import StandardScalerfrom sklearn.metrics import accuracy_score, classification_report
```

```python
cols = ['age','workclass','fnlwgt','education','education_num','marital_status','occupation',        'relationship','race','sex','capital_gain','capital_loss','hours_per_week','native_country','income']url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data'df = pd.read_csv(url, names=cols, na_values='?', skipinitialspace=True).dropna()df
```

```python
X = pd.get_dummies(df.drop('income', axis=1))y = df['income'].apply(lambda x: 1 if x == '>50K' else 0)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.2, random_state=42)scaler = StandardScaler()X_train = scaler.fit_transform(X_train).astype(np.float32)X_test = scaler.transform(X_test).astype(np.float32)y_train = y_train.to_numpy(dtype=np.float32)y_test = y_test.to_numpy(dtype=np.float32)
```

```python
# model buildingfrom tensorflow.keras.models import Sequentialfrom tensorflow.keras.layers import Dense, Inputmodel = Sequential()model.add(Input(shape=(X_train.shape[1],)))model.add(Dense(units=32, activation='relu'))model.add(Dense(units=16, activation='relu'))model.add(Dense(units=1, activation='sigmoid'))model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
# trainingmodel.fit(X_train, y_train, epochs=30, batch_size=32, validation_split=0.1)
```

```python
# predictiony_pred = model.predict(X_test)y_pred = (y_pred > 0.5).astype(int)
```

```python
# Accuracyprint("Accuracy:", accuracy_score(y_test, y_pred))print(classification_report(y_test, y_pred))
```
