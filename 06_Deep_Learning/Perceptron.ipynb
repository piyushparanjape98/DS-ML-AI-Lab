```python
import numpy as np
class perceptron:
  def __init__(self, lr = 0.1, epochs = 10):
    self.lr = lr
    self.epochs = epochs
    self.weights = None
    self.bias = 0

  def activation_fn(self, x):
    return 1 if x >= 0 else 0

  def fit(self, x, y):
    self.weights = np.zeros(x.shape[1])
    self.bias = 0

    for _ in range(self.epochs):
      for i in range(len(x)):

        #weighted sum
        total  = np.dot(x[i], self.weights) + self.bias

        #prediction
        y_pred = self.activation_fn(total)

        #error
        error = y[i] - y_pred

        #Update weights and bias
        self.weights += self.lr * error * x[i]
        self.bias += self.lr * error

  def predict(self, x):
    predict = []
    for i in x:
      total = np.dot(i, self.weights) + self.bias
      predict.append(self.activation_fn(total))
    return predict



# training the data


print("AND GATE")



X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([0,0,0,1])
model = perceptron()
model.fit(X,y)
prediction = model.predict(X)

print("Weights : ", model.weights)
print("Bias : ", model.bias)
print("\n Predictions")
for inputs, output in zip(X, prediction):
  print(f"inputs : {inputs} -> Output {output}")



print("\nOR GATE")



X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([1,1,1,0])
model = perceptron()
model.fit(X,y)
prediction = model.predict(X)

print("Weights : ", model.weights)
print("Bias : ", model.bias)
print("\n Predictions")
for inputs, output in zip(X, prediction):
  print(f"inputs : {inputs} -> Output {output}")



print("\nNAND GATE")



X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([1,1,1,0])
model = perceptron()
model.fit(X,y)
prediction = model.predict(X)

print("Weights : ", model.weights)
print("Bias : ", model.bias)
print("\n Predictions")
for inputs, output in zip(X, prediction):
  print(f"inputs : {inputs} -> Output {output}")



print("\nNOR GATE")



X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([0,0,0,1])
model = perceptron()
model.fit(X,y)
prediction = model.predict(X)

print("Weights : ", model.weights)
print("Bias : ", model.bias)
print("\n Predictions")
for inputs, output in zip(X, prediction):
  print(f"inputs : {inputs} -> Output {output}")



print("\nNOT GATE")


X = np.array([[0],[1]])
y = np.array([1,0])
model = perceptron()
model.fit(X,y)
prediction = model.predict(X)

print("Weights : ", model.weights)
print("Bias : ", model.bias)
print("\n Predictions")
for inputs, output in zip(X, prediction):
  print("inputs : ", inputs ," -> Output ", output)
```

## Why Perceptron is Better but Still Fails

The Perceptron is better than the MP model because it learns weights and bias automatically from training data instead of using fixed values. However, a single-layer perceptron can also classify only linearly separable data, so it cannot implement XOR and XNOR.
