```python
import numpy as np
```

```python
class MLP:
  def __init__(self, input_size, hidden_size, output_size, lr = 0.1, epochs = 10000):
    self.lr = lr
    self.epochs = epochs

    self.w1 = np.random.randn(input_size, hidden_size)
    self.b1 = np.zeros((1, hidden_size))

    self.w2 = np.random.randn(hidden_size, output_size)
    self.b2 = np.zeros((1, output_size))

  def Activation_fn(self, x):
    return 1 / (1 + np.exp(-x))

  def derivative_fn(self, x):
    return x * (1 - x)

  def fit(self, x, y):
    for epoch in range(self.epochs):

      # forward propogation
      hidden_input = np.dot(x, self.w1) + self.b1
      hidden_output = self.Activation_fn(hidden_input)

      final_input = np.dot(hidden_output, self.w2) + self.b2
      final_output = self.Activation_fn(final_input)

      # calculate error
      error = y - final_output

      # backpropogating
      d_final_output = error * self.derivative_fn(final_output)

      hidden_error = d_final_output.dot(self.w2.T)
      d_hidden_output = hidden_error * self.derivative_fn(hidden_output)

      # updating weights and bias
      self.w2 = self.w2 + self.lr * np.dot(hidden_output.T, d_final_output)
      self.b2 = self.b2 + self.lr * np.sum(d_final_output, axis = 0,  keepdims=True)

      self.w1 = self.w1 + self.lr * np.dot(x.T, d_hidden_output)
      self.b1 = self.b1 + self.lr * np.sum(d_hidden_output, axis = 0, keepdims=True)

  def predict(self, x):
    hidden_output = self.Activation_fn(np.dot(x, self.w1) + self.b1)
    final_output = self.Activation_fn(np.dot(hidden_output, self.w2) + self.b2)
    return (final_output >= 0.5).astype(int)
```

```python
X  = np.array([[0,0],[0,1],[1,0],[1,1]])
Y  = np.array([[0],[1],[1],[0]])

model = MLP(2, 2, 1)
model.fit(X, Y)
model.predict(X)

print("Predictions:")
for inp, pred in zip(X, model.predict(X)):
    print(f"Input: {inp} -> Output: {pred[0]}")

print(f"\nw1: {model.w1}")
print(f"\nb1: {model.b1}")
print(f"\nw2: {model.w2}")
print(f"\nb2: {model.b2}")
```
