## IQR
(Interquartile Range)

IQR is the simplest anomaly detection method — purely statistical, no model needed. It identifies outliers based on how far a point is from the middle 50% of the data.
It works by:

Finding Q1 (25th percentile) and Q3 (75th percentile)
Computing IQR = Q3 - Q1 (the spread of the middle 50%)
Setting a lower and upper fence beyond which points are considered outliers

Key idea: "If a point is unusually far from where most of the data sits, it's an outlier."



```python
import numpy as np

def detect_outliers_iqr(data, multiplier=1.5):
    Q1 = np.percentile(data, 25)
    Q3 = np.percentile(data, 75)
    IQR = Q3 - Q1

    lower_fence = Q1 - multiplier * IQR
    upper_fence = Q3 + multiplier * IQR

    outliers = [x for x in data if x < lower_fence or x > upper_fence]
    return lower_fence, upper_fence, outliers

# Example usage
data = [10, 12, 13, 14, 15, 16, 17, 18, 19, 100]
lower, upper, outliers = detect_outliers_iqr(data)

print("Lower fence:", lower)
print("Upper fence:", upper)
print("Outliers   :", outliers)
```
