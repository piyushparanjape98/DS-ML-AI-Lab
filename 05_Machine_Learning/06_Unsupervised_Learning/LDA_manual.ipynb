## LDA (Manual Implementation)

```python
import numpy as np

def lda_from_scratch(X, y, n_components):
    classes = np.unique(y)
    overall_mean = np.mean(X, axis=0)

    # Between-class scatter matrix S_B
    S_B = np.zeros((X.shape[1], X.shape[1]))
    for c in classes:
        X_c = X[y == c]
        n_c = X_c.shape[0]
        mean_c = np.mean(X_c, axis=0)
        diff = (mean_c - overall_mean).reshape(-1, 1)
        S_B += n_c * (diff @ diff.T)

    # Within-class scatter matrix S_W
    S_W = np.zeros((X.shape[1], X.shape[1]))
    for c in classes:
        X_c = X[y == c]
        mean_c = np.mean(X_c, axis=0)
        diff = X_c - mean_c
        S_W += diff.T @ diff

    # Solve eigenvalue problem
    matrix = np.linalg.inv(S_W) @ S_B
    eigenvalues, eigenvectors = np.linalg.eigh(matrix)

    # Sort descending
    sorted_idx = np.argsort(eigenvalues)[::-1]
    eigenvectors = eigenvectors[:, sorted_idx]

    # Project
    V_k = eigenvectors[:, :n_components]
    X_reduced = X @ V_k

    return X_reduced

# Example usage
from sklearn.datasets import load_iris
data = load_iris()
X_demo, y_demo = data.data, data.target

X_reduced = lda_from_scratch(X_demo, y_demo, n_components=2)

print("Original shape:", X_demo.shape)
print("Reduced shape :", X_reduced.shape)
```
