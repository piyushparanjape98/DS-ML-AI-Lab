## PCA (Manual Implementation)

```python
import numpy as np

def pca_from_scratch(X, n_components):
    # Step 1: standardize
    X_mean = X - np.mean(X, axis=0)

    # Step 2: covariance matrix
    cov_matrix = np.cov(X_mean, rowvar=False)

    # Step 3: eigenvalues and eigenvectors
    eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)

    # Step 4: sort by descending eigenvalue
    sorted_idx = np.argsort(eigenvalues)[::-1]
    eigenvalues = eigenvalues[sorted_idx]
    eigenvectors = eigenvectors[:, sorted_idx]

    # Step 5: project onto top n_components
    V_k = eigenvectors[:, :n_components]
    X_reduced = X_mean @ V_k

    # explained variance ratio
    explained = eigenvalues / eigenvalues.sum()

    return X_reduced, explained[:n_components]

# Example usage
np.random.seed(42)
X_demo = np.random.randn(100, 5)  # 100 samples, 5 features

X_reduced, explained = pca_from_scratch(X_demo, n_components=2)

print("Original shape :", X_demo.shape)
print("Reduced shape  :", X_reduced.shape)
print("Explained variance per PC:", explained.round(4))
print("Total variance explained  :", round(explained.sum(), 4))
```
